# Synthesis-Level Test Run

**Reserved. Not part of the current test run.**

The two executed notebooks in this folder cover the coding levels:

- [`01_abstractlevel_testrun.ipynb`](01_abstractlevel_testrun.ipynb): the Stage 2 abstract
  coding scheme applied to 100 PubMed records by three providers.
- [`02_fulltextlevel_testrun.ipynb`](02_fulltextlevel_testrun.ipynb): the Stage 3 full-text
  deep coding scheme applied to the retrieved open-access corpus by the same three providers.

The synthesis level (ontology-aligned semantic loading, the concept map, and the
manuscript-facing aggregation) is deliberately left until the coding schemes have been
signed off by the expert reviewers, because every synthesis input depends on the coded
values those schemes produce. Synthesizing test-run codings from three cheap models would
produce a picture of the models rather than of the literature.

The synthesis code that already exists runs from the CLI:

```bash
python -m bps_review semantic-loading
python -m bps_review build-assets
```
